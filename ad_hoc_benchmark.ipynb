```python
# ============================================================
# SINGLE COLAB CELL
# Havlíček-style ad hoc benchmark-construction-dependence audit
# using IBM hardware-estimated quantum kernels.
#
# Outputs:
#   1. Hardware quantum-kernel SVM score S_Q
#   2. Classical comparator score S_cl
#   3. eta_req = max(0, S_Q - S_cl)
#   4. Construction-side proxy eta values:
#        eta_sign_g, eta_signed_margin_bin, eta_abs_margin_bin, eta_random_control
#   5. Shortcut classifier scores using construction variables alone
#   6. Figure and CSV files
# ============================================================

!pip -q install qiskit qiskit-aer qiskit-ibm-runtime scikit-learn pandas matplotlib scipy

import os
import time
import math
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.linalg import qr

from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, balanced_accuracy_score
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.dummy import DummyClassifier

from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Statevector

try:
    from qiskit.circuit.library import zz_feature_map
    def make_feature_map(n_qubits, reps=2, entanglement="full"):
        return zz_feature_map(
            feature_dimension=n_qubits,
            reps=reps,
            entanglement=entanglement
        )
except Exception:
    from qiskit.circuit.library import ZZFeatureMap
    def make_feature_map(n_qubits, reps=2, entanglement="full"):
        return ZZFeatureMap(
            feature_dimension=n_qubits,
            reps=reps,
            entanglement=entanglement
        )

from qiskit_ibm_runtime import QiskitRuntimeService
from qiskit_ibm_runtime import SamplerV2 as IBMSamplerV2
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager


# ============================================================
# 1. USER SETTINGS
# ============================================================

# IBM credentials.
# Recommended: store these in Colab secrets or environment variables.
# You can also paste them directly below.
IBM_TOKEN = "CjF-KCAQY_inIv8WsOG-4AGw_jisXUm_MJOtaWMoqu-j"
IBM_INSTANCE = "crn:v1:bluemix:public:quantum-computing:us-east:a/70771299a825404faa640610b8510686:0876f376-35a1-4351-b85d-30e0f5c5cff6::"  # recommended

# Uncomment and paste only if not using environment variables:
# IBM_TOKEN = "paste-your-ibm-api-key-here"
# IBM_INSTANCE = "paste-your-ibm-instance-crn-here"

if IBM_TOKEN is None or IBM_INSTANCE is None:
    raise ValueError(
        "Set IBM_TOKEN and IBM_INSTANCE. In Colab, either paste them above "
        "or set environment variables IBM_QUANTUM_TOKEN and IBM_QUANTUM_INSTANCE."
    )

RANDOM_SEED = 123
rng = np.random.default_rng(RANDOM_SEED)

# Keep this small for hardware.
N_QUBITS = 2
FEATURE_MAP_REPS = 2
ENTANGLEMENT = "full"

# Ad hoc label-generation separation parameter.
# This is NOT S_Q - S_cl. It is the generator margin threshold Delta.
GENERATOR_GAP = 0.25

# Data size.
# Total train = 2 * TRAIN_PER_CLASS, total test = 2 * TEST_PER_CLASS.
TRAIN_PER_CLASS = 15
TEST_PER_CLASS = 20

# Grid used to construct candidate points in [0, 2pi)^2.
GRID_COUNT = 45

# Hardware shots per kernel circuit.
SHOTS = 1024

# Circuit batching. Smaller batches are safer on hardware.
BATCH_SIZE = 75

# Classical SVM hyperparameter grid.
C_GRID = [0.1, 1, 10, 100]


# ============================================================
# 2. BASIC UTILITIES
# ============================================================

def bind_feature_map(feature_map, x):
    """Assign feature vector x to feature-map parameters robustly."""
    params = sorted(list(feature_map.parameters), key=lambda p: str(p))
    if len(params) != len(x):
        raise ValueError(f"Expected {len(params)} features but got {len(x)}.")
    return feature_map.assign_parameters({p: float(v) for p, v in zip(params, x)}, inplace=False)


def random_unitary(dim, rng):
    """Haar-like random unitary from QR decomposition of complex Gaussian matrix."""
    z = (rng.normal(size=(dim, dim)) + 1j * rng.normal(size=(dim, dim))) / np.sqrt(2)
    q, r = qr(z)
    d = np.diag(r)
    ph = d / np.abs(d)
    q = q * ph
    return q


def z_product_diagonal(n_qubits):
    """
    Diagonal of product_i Z_i in computational basis.
    For basis state index k, eigenvalue is (-1)^(popcount(k)).
    """
    dim = 2 ** n_qubits
    diag = []
    for k in range(dim):
        parity = bin(k).count("1") % 2
        diag.append(1.0 if parity == 0 else -1.0)
    return np.diag(diag)


def statevector_for_x(feature_map, x):
    """Compute |Phi(x)> from the same feature-map circuit used in hardware kernels."""
    circ = bind_feature_map(feature_map, x)
    return np.asarray(Statevector.from_instruction(circ).data)


def expectation_score_g(feature_map, observable, x):
    """
    Latent generator score:
        g(x) = <Phi(x)| O |Phi(x)>.
    """
    psi = statevector_for_x(feature_map, x)
    val = np.vdot(psi, observable @ psi)
    return float(np.real(val))


def make_havlicek_style_dataset(
    n_qubits=2,
    reps=2,
    entanglement="full",
    gap=0.25,
    grid_count=45,
    train_per_class=5,
    test_per_class=8,
    seed=123,
):
    """
    Reimplemented ad hoc dataset so we can save the construction-side hidden variable:
        g(u) = <Phi(u)|O|Phi(u)>.
    Labels:
        y = 1 if g(u) > gap
        y = 0 if g(u) < -gap
    Points with |g(u)| <= gap are rejected.
    """
    rng_local = np.random.default_rng(seed)
    feature_map = make_feature_map(n_qubits, reps=reps, entanglement=entanglement)

    dim = 2 ** n_qubits
    zprod = z_product_diagonal(n_qubits)
    V = random_unitary(dim, rng_local)
    observable = V.conj().T @ zprod @ V

    # Grid of candidate inputs in [0, 2pi)^n.
    xvals = np.linspace(0, 2 * np.pi, grid_count, endpoint=False)
    mesh = np.meshgrid(*([xvals] * n_qubits), indexing="ij")
    X_grid = np.stack([m.reshape(-1) for m in mesh], axis=1)

    rows = []
    for x in X_grid:
        g = expectation_score_g(feature_map, observable, x)
        if g > gap:
            y = 1
            accepted = True
        elif g < -gap:
            y = 0
            accepted = True
        else:
            y = -1
            accepted = False

        rows.append({
            "x0": x[0],
            "x1": x[1] if n_qubits > 1 else np.nan,
            "g": g,
            "abs_g": abs(g),
            "accepted": accepted,
            "y": y
        })

    all_df = pd.DataFrame(rows)
    acc_df = all_df[all_df["accepted"]].copy()

    n_needed = train_per_class + test_per_class
    counts = acc_df["y"].value_counts().to_dict()
    if counts.get(0, 0) < n_needed or counts.get(1, 0) < n_needed:
        raise RuntimeError(
            f"Not enough accepted samples per class. Counts={counts}. "
            "Try lower GENERATOR_GAP or larger GRID_COUNT."
        )

    selected_parts = []
    for cls in [0, 1]:
        cls_df = acc_df[acc_df["y"] == cls].sample(
            n=n_needed,
            replace=False,
            random_state=seed + cls
        ).copy()
        cls_df["split"] = ["train"] * train_per_class + ["test"] * test_per_class
        selected_parts.append(cls_df)

    selected = pd.concat(selected_parts, ignore_index=True).sample(
        frac=1,
        random_state=seed
    ).reset_index(drop=True)

    feature_cols = [f"x{i}" for i in range(n_qubits)]
    train_df = selected[selected["split"] == "train"].copy().reset_index(drop=True)
    test_df = selected[selected["split"] == "test"].copy().reset_index(drop=True)

    X_train = train_df[feature_cols].values
    y_train = train_df["y"].astype(int).values
    X_test = test_df[feature_cols].values
    y_test = test_df["y"].astype(int).values

    return feature_map, observable, all_df, selected, train_df, test_df, X_train, y_train, X_test, y_test


def total_variation_from_counts(p_counts, q_counts, categories):
    """TV distance between two categorical empirical distributions."""
    p_total = sum(p_counts.get(c, 0) for c in categories)
    q_total = sum(q_counts.get(c, 0) for c in categories)

    if p_total == 0 or q_total == 0:
        return np.nan

    tv = 0.0
    for c in categories:
        p = p_counts.get(c, 0) / p_total
        q = q_counts.get(c, 0) / q_total
        tv += abs(p - q)
    return 0.5 * tv


def eta_for_categorical_lambda(df, lambda_col, y_col="y"):
    """
    eta_lambda = max_y TV[p(lambda|y), p0(lambda)].
    """
    categories = sorted(df[lambda_col].dropna().unique().tolist())
    overall_counts = df[lambda_col].value_counts().to_dict()

    eta_by_y = {}
    for yval in sorted(df[y_col].unique()):
        sub_counts = df[df[y_col] == yval][lambda_col].value_counts().to_dict()
        eta_by_y[yval] = total_variation_from_counts(sub_counts, overall_counts, categories)

    return float(np.nanmax(list(eta_by_y.values()))), eta_by_y


def majority_shortcut_predict(train_df, test_df, lambda_col, y_col="y"):
    """
    Classifier that uses lambda alone:
        for each lambda category, predict the majority training label.
    Unseen lambda categories use global training majority.
    """
    global_majority = int(train_df[y_col].mode().iloc[0])
    mapping = {}
    for val, sub in train_df.groupby(lambda_col):
        mapping[val] = int(sub[y_col].mode().iloc[0])

    preds = []
    for val in test_df[lambda_col].values:
        preds.append(mapping.get(val, global_majority))
    return np.array(preds, dtype=int), mapping


# ============================================================
# 3. IBM HARDWARE SETUP
# ============================================================

print("Connecting to IBM Quantum Runtime...")
service = QiskitRuntimeService(
    channel="ibm_quantum_platform",
    token=IBM_TOKEN,
    instance=IBM_INSTANCE,
)

backend = service.least_busy(
    operational=True,
    simulator=False,
    min_num_qubits=N_QUBITS
)

print("Selected backend:", backend.name)
print("Active instance:", service.active_instance())

pass_manager = generate_preset_pass_manager(
    backend=backend,
    optimization_level=1
)

sampler = IBMSamplerV2(mode=backend)
sampler.options.default_shots = SHOTS


# ============================================================
# 4. HARDWARE KERNEL CIRCUITS
# ============================================================

def fidelity_kernel_circuit(feature_map, x, xp):
    """
    Circuit for estimating
        K_Q(x,xp) = |<Phi(x)|Phi(xp)>|^2.
    It prepares U_phi(x)|0>, applies U_phi(xp)^dagger, and measures.
    The all-zero probability is the kernel element.
    """
    n = feature_map.num_qubits
    circ_x = bind_feature_map(feature_map, x)
    circ_xp = bind_feature_map(feature_map, xp)

    qc = QuantumCircuit(n)
    qc.compose(circ_x, inplace=True)
    qc.compose(circ_xp.inverse(), inplace=True)
    qc.measure_all()
    return qc


def extract_counts_from_sampler_result(pub_result):
    """
    Robust extraction of counts from SamplerV2 PubResult.
    Usually: pub_result.data.meas.get_counts()
    """
    # Common case
    try:
        return pub_result.data.meas.get_counts()
    except Exception:
        pass

    # Fallback: find first classical register-like object with get_counts()
    try:
        data = pub_result.data
        for name in dir(data):
            if name.startswith("_"):
                continue
            obj = getattr(data, name)
            if hasattr(obj, "get_counts"):
                return obj.get_counts()
    except Exception:
        pass

    raise RuntimeError(
        "Could not extract counts from SamplerV2 result. "
        "Inspect result[0].data to adjust extraction."
    )


def zero_probability_from_counts(counts, n_qubits, shots):
    zero = "0" * n_qubits
    count0 = 0
    for key, val in counts.items():
        key_clean = str(key).replace(" ", "")
        if key_clean == zero:
            count0 += val
    return count0 / shots


def run_hardware_circuits(circuits, batch_size=75, shots=1024):
    """
    Transpile and run circuits on IBM hardware using SamplerV2.
    """
    all_counts = []
    n_total = len(circuits)

    for start in range(0, n_total, batch_size):
        batch = circuits[start:start + batch_size]
        batch_isa = [pass_manager.run(c) for c in batch]

        print(f"Submitting circuits {start + 1}-{start + len(batch)} of {n_total}...")
        job = sampler.run(batch_isa, shots=shots)
        print("  Job ID:", job.job_id())
        result = job.result()

        for k in range(len(batch_isa)):
            counts = extract_counts_from_sampler_result(result[k])
            all_counts.append(counts)

        print(f"  Completed {start + len(batch)} / {n_total}")
        time.sleep(2)

    return all_counts


def estimate_hardware_kernel(X_left, X_right, feature_map, symmetric=False, shots=1024):
    """
    Estimate quantum fidelity kernel matrix from IBM hardware.
    """
    circuits = []
    pairs = []

    if symmetric:
        n = len(X_left)
        for i in range(n):
            for j in range(i, n):
                circuits.append(fidelity_kernel_circuit(feature_map, X_left[i], X_left[j]))
                pairs.append((i, j))
    else:
        for i in range(len(X_left)):
            for j in range(len(X_right)):
                circuits.append(fidelity_kernel_circuit(feature_map, X_left[i], X_right[j]))
                pairs.append((i, j))

    counts_list = run_hardware_circuits(
        circuits,
        batch_size=BATCH_SIZE,
        shots=shots
    )

    if symmetric:
        n = len(X_left)
        K = np.zeros((n, n), dtype=float)
        for counts, (i, j) in zip(counts_list, pairs):
            p0 = zero_probability_from_counts(counts, feature_map.num_qubits, shots)
            K[i, j] = p0
            K[j, i] = p0
        # Physically, K(x,x)=1 ideally. Hardware noise may reduce it, but SVM kernels are more stable with diag=1.
        np.fill_diagonal(K, 1.0)
        K = 0.5 * (K + K.T)
    else:
        K = np.zeros((len(X_left), len(X_right)), dtype=float)
        for counts, (i, j) in zip(counts_list, pairs):
            p0 = zero_probability_from_counts(counts, feature_map.num_qubits, shots)
            K[i, j] = p0

    return np.clip(K, 0.0, 1.0)


# ============================================================
# 5. GENERATE AD HOC DATA WITH LATENT CONSTRUCTION VARIABLES
# ============================================================

feature_map, observable, grid_df, selected_df, train_df, test_df, X_train, y_train, X_test, y_test = make_havlicek_style_dataset(
    n_qubits=N_QUBITS,
    reps=FEATURE_MAP_REPS,
    entanglement=ENTANGLEMENT,
    gap=GENERATOR_GAP,
    grid_count=GRID_COUNT,
    train_per_class=TRAIN_PER_CLASS,
    test_per_class=TEST_PER_CLASS,
    seed=RANDOM_SEED,
)

# Construction-side hidden variables.
def add_construction_lambdas(df, rng_seed=123):
    out = df.copy()
    out["lambda_sign_g"] = np.where(out["g"] >= 0, "g_positive", "g_negative")

    # Signed-margin bins. These are coarse construction variables.
    signed_bins = [-1.0, -0.75, -0.50, -GENERATOR_GAP, GENERATOR_GAP, 0.50, 0.75, 1.0]
    signed_labels = ["neg_high", "neg_mid", "neg_low", "gap_region", "pos_low", "pos_mid", "pos_high"]
    out["lambda_signed_margin_bin"] = pd.cut(
        out["g"],
        bins=signed_bins,
        labels=signed_labels,
        include_lowest=True
    ).astype(str)

    # Magnitude bins: weak/medium/strong margin.
    try:
        out["lambda_abs_margin_bin"] = pd.qcut(
            out["abs_g"],
            q=3,
            labels=["abs_low", "abs_mid", "abs_high"],
            duplicates="drop"
        ).astype(str)
    except Exception:
        out["lambda_abs_margin_bin"] = pd.cut(
            out["abs_g"],
            bins=3,
            labels=["abs_low", "abs_mid", "abs_high"],
            include_lowest=True
        ).astype(str)

    rng_local = np.random.default_rng(rng_seed)
    out["lambda_random_control"] = rng_local.choice(["r0", "r1"], size=len(out))
    return out

train_df = add_construction_lambdas(train_df, RANDOM_SEED + 1)
test_df = add_construction_lambdas(test_df, RANDOM_SEED + 2)
selected_df = pd.concat([train_df, test_df], ignore_index=True)

print("\nDataset summary")
print("Train counts:", pd.Series(y_train).value_counts().to_dict())
print("Test counts:", pd.Series(y_test).value_counts().to_dict())
print("Selected data:")
display(selected_df.head())


# ============================================================
# 6. QUANTUM HARDWARE KERNEL SVM
# ============================================================

print("\nEstimating TRAIN quantum kernel on IBM hardware...")
K_train = estimate_hardware_kernel(
    X_train,
    X_train,
    feature_map,
    symmetric=True,
    shots=SHOTS
)

print("\nEstimating TEST quantum kernel on IBM hardware...")
K_test = estimate_hardware_kernel(
    X_test,
    X_train,
    feature_map,
    symmetric=False,
    shots=SHOTS
)

# Tune C for precomputed quantum kernel using tiny inner CV.
min_class_count = int(np.min(np.bincount(y_train)))
n_splits = max(2, min(3, min_class_count))

best_q_C = None
best_q_cv = -np.inf

if n_splits >= 2:
    cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=RANDOM_SEED)
    idx = np.arange(len(y_train))

    for C in C_GRID:
        scores = []
        for tr, va in cv.split(idx, y_train):
            K_tr = K_train[np.ix_(tr, tr)]
            K_va = K_train[np.ix_(va, tr)]
            clf = SVC(kernel="precomputed", C=C)
            clf.fit(K_tr, y_train[tr])
            pred = clf.predict(K_va)
            scores.append(balanced_accuracy_score(y_train[va], pred))
        mean_score = np.mean(scores)
        if mean_score > best_q_cv:
            best_q_cv = mean_score
            best_q_C = C
else:
    best_q_C = 1.0
    best_q_cv = np.nan

q_clf = SVC(kernel="precomputed", C=best_q_C)
q_clf.fit(K_train, y_train)
y_pred_q = q_clf.predict(K_test)

S_Q = accuracy_score(y_test, y_pred_q)
BA_Q = balanced_accuracy_score(y_test, y_pred_q)


# ============================================================
# 7. CLASSICAL COMPARATOR SVMs
# ============================================================

classical_models = {
    "linear": (
        Pipeline([("scaler", StandardScaler()), ("svc", SVC(kernel="linear"))]),
        {"svc__C": C_GRID}
    ),
    "poly": (
        Pipeline([("scaler", StandardScaler()), ("svc", SVC(kernel="poly"))]),
        {
            "svc__C": C_GRID,
            "svc__degree": [2, 3, 4],
            "svc__gamma": ["scale", 0.1, 1.0],
            "svc__coef0": [0.0, 1.0]
        }
    ),
    "rbf": (
        Pipeline([("scaler", StandardScaler()), ("svc", SVC(kernel="rbf"))]),
        {
            "svc__C": C_GRID,
            "svc__gamma": ["scale", 0.01, 0.1, 1.0, 10.0]
        }
    )
}

best_cl_name = None
best_cl_fit = None
best_cl_cv = -np.inf
all_cl_fits = {}

cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=RANDOM_SEED)

for name, (pipe, grid) in classical_models.items():
    gs = GridSearchCV(
        pipe,
        grid,
        scoring="balanced_accuracy",
        cv=cv,
        n_jobs=-1,
        refit=True
    )
    gs.fit(X_train, y_train)
    all_cl_fits[name] = gs
    if gs.best_score_ > best_cl_cv:
        best_cl_cv = gs.best_score_
        best_cl_name = name
        best_cl_fit = gs

y_pred_cl = best_cl_fit.predict(X_test)
S_cl = accuracy_score(y_test, y_pred_cl)
BA_cl = balanced_accuracy_score(y_test, y_pred_cl)

eta_req = max(0.0, S_Q - S_cl)


# ============================================================
# 8. BENCHMARK-CONSTRUCTION DEPENDENCE AUDIT
# ============================================================

lambda_cols = [
    "lambda_sign_g",
    "lambda_signed_margin_bin",
    "lambda_abs_margin_bin",
    "lambda_random_control"
]

audit_rows = []
shortcut_scores = {}

for lam in lambda_cols:
    eta_full, eta_by_class_full = eta_for_categorical_lambda(selected_df, lam, y_col="y")
    eta_test, eta_by_class_test = eta_for_categorical_lambda(test_df, lam, y_col="y")

    y_pred_shortcut, mapping = majority_shortcut_predict(train_df, test_df, lam, y_col="y")
    S_lam = accuracy_score(y_test, y_pred_shortcut)
    BA_lam = balanced_accuracy_score(y_test, y_pred_shortcut)

    audit_rows.append({
        "lambda_variable": lam,
        "eta_full_train_plus_test": eta_full,
        "eta_test_only": eta_test,
        "shortcut_score": S_lam,
        "shortcut_balanced_score": BA_lam,
        "eta_req_from_quantum_minus_classical": eta_req,
        "shortcut_mapping": str(mapping),
        "eta_by_class_full": str(eta_by_class_full),
        "eta_by_class_test": str(eta_by_class_test)
    })
    shortcut_scores[lam] = S_lam

audit_df = pd.DataFrame(audit_rows)

summary = pd.DataFrame([{
    "backend": backend.name,
    "n_qubits": N_QUBITS,
    "feature_map_reps": FEATURE_MAP_REPS,
    "entanglement": ENTANGLEMENT,
    "generator_gap_Delta": GENERATOR_GAP,
    "shots": SHOTS,
    "train_per_class": TRAIN_PER_CLASS,
    "test_per_class": TEST_PER_CLASS,
    "S_Q_hardware_quantum_kernel": S_Q,
    "BA_Q_hardware_quantum_kernel": BA_Q,
    "q_best_C": best_q_C,
    "q_inner_cv_balanced_accuracy": best_q_cv,
    "best_classical_kernel": best_cl_name,
    "S_cl": S_cl,
    "BA_cl": BA_cl,
    "best_classical_inner_cv_balanced_accuracy": best_cl_cv,
    "delta_Q_minus_cl": S_Q - S_cl,
    "eta_req": eta_req,
    "S_shortcut_sign_g": shortcut_scores["lambda_sign_g"],
    "S_shortcut_signed_margin_bin": shortcut_scores["lambda_signed_margin_bin"],
    "S_shortcut_abs_margin_bin": shortcut_scores["lambda_abs_margin_bin"],
    "S_shortcut_random_control": shortcut_scores["lambda_random_control"],
}])

paired_predictions = pd.DataFrame({
    "y_true": y_test,
    "y_pred_q": y_pred_q,
    "y_pred_cl": y_pred_cl,
    "q_correct": (y_pred_q == y_test).astype(int),
    "cl_correct": (y_pred_cl == y_test).astype(int),
    "paired_correct_diff": (y_pred_q == y_test).astype(int) - (y_pred_cl == y_test).astype(int),
    "g": test_df["g"].values,
    "abs_g": test_df["abs_g"].values,
    "lambda_sign_g": test_df["lambda_sign_g"].values,
    "lambda_signed_margin_bin": test_df["lambda_signed_margin_bin"].values,
    "lambda_abs_margin_bin": test_df["lambda_abs_margin_bin"].values,
    "lambda_random_control": test_df["lambda_random_control"].values,
})


# ============================================================
# 9. SAVE OUTPUTS
# ============================================================

summary.to_csv("adhoc_hardware_construction_dependence_summary.csv", index=False)
audit_df.to_csv("adhoc_hardware_construction_dependence_audit.csv", index=False)
selected_df.to_csv("adhoc_hardware_selected_dataset_with_latent_variables.csv", index=False)
paired_predictions.to_csv("adhoc_hardware_paired_predictions.csv", index=False)

np.save("adhoc_hardware_K_train.npy", K_train)
np.save("adhoc_hardware_K_test.npy", K_test)

print("\nMain summary")
display(summary)

print("\nConstruction-dependence audit")
display(audit_df)

print("\nPaired predictions")
display(paired_predictions)


# ============================================================
# 10. FIGURE: FOUR-PANEL AUDIT
# ============================================================

fig, axes = plt.subplots(2, 2, figsize=(12, 9))

# Panel A: dataset scatter
ax = axes[0, 0]
for cls, marker in [(0, "o"), (1, "s")]:
    tmp = selected_df[selected_df["y"] == cls]
    ax.scatter(tmp["x0"], tmp["x1"], label=f"class {cls}", alpha=0.85, marker=marker)
ax.set_xlabel(r"$u_1$")
ax.set_ylabel(r"$u_2$")
ax.set_title("A. Ad hoc benchmark samples")
ax.legend(frameon=False)

# Panel B: latent generator score histogram
ax = axes[0, 1]
bins = np.linspace(-1, 1, 18)
ax.hist(selected_df[selected_df["y"] == 0]["g"], bins=bins, alpha=0.65, label="class 0")
ax.hist(selected_df[selected_df["y"] == 1]["g"], bins=bins, alpha=0.65, label="class 1")
ax.axvline(-GENERATOR_GAP, linestyle="--", linewidth=1)
ax.axvline(GENERATOR_GAP, linestyle="--", linewidth=1)
ax.set_xlabel(r"latent generator score $g(u)$")
ax.set_ylabel("count")
ax.set_title("B. Construction score by class")
ax.legend(frameon=False)

# Panel C: eta values for construction variables
ax = axes[1, 0]
eta_plot = audit_df.copy()
labels = eta_plot["lambda_variable"].str.replace("lambda_", "", regex=False)
x = np.arange(len(eta_plot))
ax.bar(x, eta_plot["eta_test_only"])
ax.axhline(eta_req, linestyle="--", linewidth=1, label=r"$\eta_{\rm req}$")
ax.set_xticks(x)
ax.set_xticklabels(labels, rotation=25, ha="right")
ax.set_ylabel(r"$\widehat{\eta}_{\lambda}$")
ax.set_title("C. Benchmark-construction dependence")
ax.legend(frameon=False)

# Panel D: score comparison
ax = axes[1, 1]
score_names = [
    "classical\nSVM",
    "hardware\nQ kernel",
    "shortcut\nsign(g)",
    "shortcut\nsigned bin",
    "shortcut\nabs bin",
    "random\ncontrol"
]
score_vals = [
    S_cl,
    S_Q,
    shortcut_scores["lambda_sign_g"],
    shortcut_scores["lambda_signed_margin_bin"],
    shortcut_scores["lambda_abs_margin_bin"],
    shortcut_scores["lambda_random_control"],
]
x = np.arange(len(score_vals))
ax.bar(x, score_vals)
ax.set_ylim(0, 1.05)
ax.set_xticks(x)
ax.set_xticklabels(score_names, rotation=25, ha="right")
ax.set_ylabel("test accuracy")
ax.set_title("D. Quantum/classical/shortcut scores")

plt.tight_layout()
plt.savefig("adhoc_hardware_construction_dependence_figure.pdf", bbox_inches="tight")
plt.savefig("adhoc_hardware_construction_dependence_figure.png", dpi=300, bbox_inches="tight")
plt.show()

print("\nSaved files:")
print("  adhoc_hardware_construction_dependence_summary.csv")
print("  adhoc_hardware_construction_dependence_audit.csv")
print("  adhoc_hardware_selected_dataset_with_latent_variables.csv")
print("  adhoc_hardware_paired_predictions.csv")
print("  adhoc_hardware_K_train.npy")
print("  adhoc_hardware_K_test.npy")
print("  adhoc_hardware_construction_dependence_figure.pdf")
print("  adhoc_hardware_construction_dependence_figure.png")

print("\nInterpretation:")
print(f"  S_Q  = {S_Q:.3f}")
print(f"  S_cl = {S_cl:.3f}")
print(f"  eta_req = max(0, S_Q - S_cl) = {eta_req:.3f}")
print("  Compare eta_req to eta_test_only for construction variables.")
print("  If eta_lambda > eta_req, that construction-side variable is strong enough,")
print("  in principle, to explain the apparent quantum-kernel excess.")
```

```python
# ============================================================
# Classical SVM with access to construction-side hidden variables
# Uses saved files from the hardware benchmark run
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.metrics import accuracy_score, balanced_accuracy_score

# ------------------------------------------------------------
# 1. Load saved hardware benchmark outputs
# ------------------------------------------------------------

DATA_FILE = "adhoc_hardware_selected_dataset_with_latent_variables.csv"
SUMMARY_FILE = "adhoc_hardware_construction_dependence_summary.csv"

df = pd.read_csv(DATA_FILE)
summary = pd.read_csv(SUMMARY_FILE)

train_df = df[df["split"] == "train"].copy().reset_index(drop=True)
test_df = df[df["split"] == "test"].copy().reset_index(drop=True)

y_train = train_df["y"].astype(int).values
y_test = test_df["y"].astype(int).values

S_Q = float(summary["S_Q_hardware_quantum_kernel"].iloc[0])
S_cl_original = float(summary["S_cl"].iloc[0])
eta_req = float(summary["eta_req"].iloc[0])

print("Original hardware benchmark:")
print(f"  S_Q       = {S_Q:.4f}")
print(f"  S_cl      = {S_cl_original:.4f}")
print(f"  eta_req   = {eta_req:.4f}")
print()
print("Train class counts:", train_df["y"].value_counts().to_dict())
print("Test class counts: ", test_df["y"].value_counts().to_dict())

# ------------------------------------------------------------
# 2. Robust OneHotEncoder helper
# ------------------------------------------------------------

def make_onehot():
    # sklearn changed sparse -> sparse_output in newer versions
    try:
        return OneHotEncoder(handle_unknown="ignore", sparse_output=False)
    except TypeError:
        return OneHotEncoder(handle_unknown="ignore", sparse=False)

# ------------------------------------------------------------
# 3. General SVM fitting function
# ------------------------------------------------------------

def fit_svm_with_columns(train_df, test_df, numeric_cols=None, categorical_cols=None, label_col="y"):
    """
    Fit a classical SVM using selected numeric and categorical columns.

    This is intentionally allowed to include construction-side hidden variables.
    That makes it a benchmark-dependent classical surrogate, not the fair S_cl comparator.
    """
    numeric_cols = numeric_cols or []
    categorical_cols = categorical_cols or []

    X_train = train_df[numeric_cols + categorical_cols]
    X_test = test_df[numeric_cols + categorical_cols]

    y_train = train_df[label_col].astype(int).values
    y_test = test_df[label_col].astype(int).values

    transformers = []

    if numeric_cols:
        transformers.append(("num", StandardScaler(), numeric_cols))

    if categorical_cols:
        transformers.append(("cat", make_onehot(), categorical_cols))

    preprocessor = ColumnTransformer(transformers)

    pipe = Pipeline([
        ("prep", preprocessor),
        ("svc", SVC())
    ])

    # Keep grid modest because the training set is tiny.
    param_grid = [
        {
            "svc__kernel": ["linear"],
            "svc__C": [0.1, 1, 10, 100],
        },
        {
            "svc__kernel": ["rbf"],
            "svc__C": [0.1, 1, 10, 100],
            "svc__gamma": ["scale", 0.1, 1.0, 10.0],
        },
        {
            "svc__kernel": ["poly"],
            "svc__C": [0.1, 1, 10, 100],
            "svc__degree": [2, 3],
            "svc__gamma": ["scale", 0.1, 1.0],
            "svc__coef0": [0.0, 1.0],
        },
    ]

    min_class_count = int(np.min(np.bincount(y_train)))
    n_splits = max(2, min(3, min_class_count))

    cv = StratifiedKFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=123
    )

    gs = GridSearchCV(
        pipe,
        param_grid,
        scoring="balanced_accuracy",
        cv=cv,
        n_jobs=-1,
        refit=True
    )

    gs.fit(X_train, y_train)
    pred = gs.predict(X_test)

    out = {
        "accuracy": accuracy_score(y_test, pred),
        "balanced_accuracy": balanced_accuracy_score(y_test, pred),
        "best_params": gs.best_params_,
        "best_cv_balanced_accuracy": gs.best_score_,
        "pred": pred,
    }

    return out

# ------------------------------------------------------------
# 4. Define official features and hidden variables
# ------------------------------------------------------------

official_features = ["x0", "x1"]

hidden_specs = {
    "sign_g": {
        "numeric": [],
        "categorical": ["lambda_sign_g"],
        "description": "sign of latent generator score"
    },
    "signed_margin_bin": {
        "numeric": [],
        "categorical": ["lambda_signed_margin_bin"],
        "description": "coarse signed generator-margin bin"
    },
    "abs_margin_bin": {
        "numeric": [],
        "categorical": ["lambda_abs_margin_bin"],
        "description": "coarse absolute-margin bin"
    },
    "random_control": {
        "numeric": [],
        "categorical": ["lambda_random_control"],
        "description": "random control variable"
    },
    "continuous_g": {
        "numeric": ["g"],
        "categorical": [],
        "description": "continuous latent generator score"
    },
    "continuous_abs_g": {
        "numeric": ["abs_g"],
        "categorical": [],
        "description": "continuous absolute generator margin"
    },
}

# ------------------------------------------------------------
# 5. Fit baseline official-feature classical SVM
# ------------------------------------------------------------

baseline = fit_svm_with_columns(
    train_df,
    test_df,
    numeric_cols=official_features,
    categorical_cols=[],
)

rows = []

rows.append({
    "model": "official_features_only",
    "input_type": "u only",
    "hidden_variable": "none",
    "accuracy": baseline["accuracy"],
    "balanced_accuracy": baseline["balanced_accuracy"],
    "beats_quantum": baseline["accuracy"] >= S_Q,
    "excess_over_quantum": baseline["accuracy"] - S_Q,
    "best_params": str(baseline["best_params"]),
})

# ------------------------------------------------------------
# 6. Fit hidden-only and official+hidden SVMs
# ------------------------------------------------------------

for name, spec in hidden_specs.items():
    # Hidden variable only
    hidden_only = fit_svm_with_columns(
        train_df,
        test_df,
        numeric_cols=spec["numeric"],
        categorical_cols=spec["categorical"],
    )

    rows.append({
        "model": f"hidden_only__{name}",
        "input_type": "lambda only",
        "hidden_variable": name,
        "accuracy": hidden_only["accuracy"],
        "balanced_accuracy": hidden_only["balanced_accuracy"],
        "beats_quantum": hidden_only["accuracy"] >= S_Q,
        "excess_over_quantum": hidden_only["accuracy"] - S_Q,
        "best_params": str(hidden_only["best_params"]),
    })

    # Official features plus hidden variable
    official_plus_hidden = fit_svm_with_columns(
        train_df,
        test_df,
        numeric_cols=official_features + spec["numeric"],
        categorical_cols=spec["categorical"],
    )

    rows.append({
        "model": f"official_plus_hidden__{name}",
        "input_type": "u + lambda",
        "hidden_variable": name,
        "accuracy": official_plus_hidden["accuracy"],
        "balanced_accuracy": official_plus_hidden["balanced_accuracy"],
        "beats_quantum": official_plus_hidden["accuracy"] >= S_Q,
        "excess_over_quantum": official_plus_hidden["accuracy"] - S_Q,
        "best_params": str(official_plus_hidden["best_params"]),
    })

augmented_results = pd.DataFrame(rows)

# ------------------------------------------------------------
# 7. Load eta values from construction-dependence audit if available
# ------------------------------------------------------------

audit_file = "adhoc_hardware_construction_dependence_audit.csv"

try:
    audit = pd.read_csv(audit_file)
    eta_map = {
        "sign_g": float(audit.loc[audit["lambda_variable"] == "lambda_sign_g", "eta_test_only"].iloc[0]),
        "signed_margin_bin": float(audit.loc[audit["lambda_variable"] == "lambda_signed_margin_bin", "eta_test_only"].iloc[0]),
        "abs_margin_bin": float(audit.loc[audit["lambda_variable"] == "lambda_abs_margin_bin", "eta_test_only"].iloc[0]),
        "random_control": float(audit.loc[audit["lambda_variable"] == "lambda_random_control", "eta_test_only"].iloc[0]),
        "continuous_g": np.nan,
        "continuous_abs_g": np.nan,
        "none": 0.0,
    }
    augmented_results["eta_lambda_test_only"] = augmented_results["hidden_variable"].map(eta_map)
    augmented_results["eta_req"] = eta_req
    augmented_results["eta_lambda_exceeds_eta_req"] = augmented_results["eta_lambda_test_only"] > eta_req
except Exception as e:
    print("Could not load eta audit file:", e)
    augmented_results["eta_lambda_test_only"] = np.nan
    augmented_results["eta_req"] = eta_req
    augmented_results["eta_lambda_exceeds_eta_req"] = np.nan

# ------------------------------------------------------------
# 8. Display and save results
# ------------------------------------------------------------

augmented_results = augmented_results.sort_values(
    by=["accuracy", "balanced_accuracy"],
    ascending=False
).reset_index(drop=True)

display(augmented_results)

augmented_results.to_csv("adhoc_classical_svm_with_hidden_variables.csv", index=False)

print("Saved: adhoc_classical_svm_with_hidden_variables.csv")

# ------------------------------------------------------------
# 9. Plot comparison
# ------------------------------------------------------------

plot_df = augmented_results.copy()
plot_df["short_label"] = plot_df["model"].str.replace("official_plus_hidden__", "u+", regex=False)
plot_df["short_label"] = plot_df["short_label"].str.replace("hidden_only__", "lambda:", regex=False)
plot_df["short_label"] = plot_df["short_label"].str.replace("official_features_only", "u only", regex=False)

plt.figure(figsize=(11, 5.5))

x = np.arange(len(plot_df))
plt.bar(x, plot_df["accuracy"])
plt.axhline(S_Q, linestyle="--", linewidth=1.5, label=r"hardware quantum kernel $S_Q$")
plt.axhline(S_cl_original, linestyle=":", linewidth=1.5, label=r"original classical comparator $S_{\rm cl}$")

plt.xticks(x, plot_df["short_label"], rotation=35, ha="right")
plt.ylim(0, 1.05)
plt.ylabel("test accuracy")
plt.xlabel("classical SVM input")
plt.title("Classical SVM with access to construction-side hidden variables")
plt.legend(frameon=False)
plt.tight_layout()

plt.savefig("adhoc_classical_svm_hidden_variable_access.pdf", bbox_inches="tight")
plt.savefig("adhoc_classical_svm_hidden_variable_access.png", dpi=300, bbox_inches="tight")
plt.show()

# ------------------------------------------------------------
# 10. Plain-language summary
# ------------------------------------------------------------

best = augmented_results.iloc[0]

print("\nInterpretation:")
print(f"  Hardware quantum-kernel score S_Q = {S_Q:.3f}")
print(f"  Original loophole-free classical score S_cl = {S_cl_original:.3f}")
print(f"  eta_req = max(0, S_Q - S_cl) = {eta_req:.3f}")
print()
print(f"  Best hidden-variable-enabled classical SVM: {best['model']}")
print(f"  Accuracy = {best['accuracy']:.3f}")
print(f"  Beats or matches quantum? {bool(best['beats_quantum'])}")
print()
print("If a hidden-variable-enabled classical SVM matches or exceeds S_Q,")
print("then the construction-side hidden variable is operationally sufficient")
print("to reproduce the apparent quantum-kernel advantage.")
```
